```python
summary = pd.DataFrame({
    "Metric": [
        "Total Students",
        "Average Final Grade",
        "Highest Final Grade",
        "Lowest Final Grade",
        "Average Absences",
        "Average Study Time"
    ],
    "Value": [
        len(student),
        round(student["G3"].mean(), 2),
        student["G3"].max(),
        student["G3"].min(),
        round(student["absences"].mean(), 2),
        round(student["studytime"].mean(), 2)
    ]
})

summary
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# 1. Final Grade Distribution
axes[0, 0].hist(student["G3"], bins=10)
axes[0, 0].set_title("Final Grade Distribution")
axes[0, 0].set_xlabel("Final Grade (G3)")
axes[0, 0].set_ylabel("Number of Students")

# 2. Absences vs Final Grade
axes[0, 1].scatter(student["absences"], student["G3"])
axes[0, 1].set_title("Absences vs Final Grade")
axes[0, 1].set_xlabel("Absences")
axes[0, 1].set_ylabel("Final Grade (G3)")

# 3. Study Time vs Average Grade
student.groupby("studytime")["G3"].mean().plot(
    kind="bar",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Study Time vs Average Grade")
axes[1, 0].set_xlabel("Study Time")
axes[1, 0].set_ylabel("Average G3")

# 4. Failures vs Average Grade
student.groupby("failures")["G3"].mean().plot(
    kind="bar",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Failures vs Average Grade")
axes[1, 1].set_xlabel("Number of Failures")
axes[1, 1].set_ylabel("Average G3")

plt.tight_layout()
plt.show()
```
